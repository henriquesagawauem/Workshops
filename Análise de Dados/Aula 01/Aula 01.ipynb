**Filtre os pedidos com desconto acima de 30% (`Discount > 0.3`) E da categoria `Furniture`.  
Quantos pedidos atendem às duas condições simultaneamente?  
Qual é o lucro médio desses pedidos? Eles são lucrativos?**

> 💡 *Dica: depois de filtrar, use `['Profit'].mean()` para a média de lucro.*


```python
df_furniture = df['Category'] == 'Furniture'
df_discount = df['Discount'] > 0.3

df_furn_disc = df[df_furniture & df_discount]

display(len(df_furn_disc))
print(df_furn_disc['Profit'].mean())

if df_furn_disc['Profit'].mean() > 0:
  print("Dá")
else:
  print("Não Dá")
```

**Use `isin()` para filtrar pedidos dos segmentos `Corporate` e `Home Office`.
